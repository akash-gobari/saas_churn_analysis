```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
# Data
url = "https://docs.google.com/spreadsheets/d/1oRnkMN6AdVQNVJEnBs_krm6Qaj8MEA1UnvhykvCzdkE/export?format=csv&gid=880221893"
df=pd.read_csv(url)
df.head()
```

## 1. Datset Exploration

```python
# Total number of records
print("Total Number of Records:",df.shape[0])
print("Total Number of Columns:",df.shape[1])
print("_"*100)
# Data type of variables
print("Data type of variables:")
df.dtypes
```

```python
# Summary statistics of numerical columns
df.describe().T
```

## 2. Data Preparation

```python
# Checking for missing values
print("Total Missing Values in dataset:",df.isna().sum().sum()) 

# Check for duplicate recors
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Converting signup_Date from string to datetime
df['signup_date'] = pd.to_datetime(df['signup_date'])

print(df['signup_date'].dtype)
```

```python
# Checking Categorical Values
print(df['plan_type'].value_counts())
print("_"*50)
print(df['churn'].value_counts())
```

```python
# Converting churn values from yes/no -> 1/0
df['churn'] = df['churn'].map({"No": 0, "Yes":1})

df.head()
```

## 3. Customer Engagement Analysis

```python
# Montly Usage
print("Average of weekly usage hours:", df['avg_weekly_usage_hours'].mean())
print("Median of weekly usage hours:", df['avg_weekly_usage_hours'].median())
```

```python
from IPython.display import display, Markdown
# Engagement by subscription plan
display(Markdown("#### Weekly Usage Hours by Subscription Plan"))
display(df.groupby('plan_type')['avg_weekly_usage_hours'].agg(
    ['count', 'mean', 'median', 'min', 'max']
))
```

```python
# Usage vs Tenure
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='tenure_months',
    y='avg_weekly_usage_hours',
    hue='churn'
)
plt.title('Weekly Usage vs Customer Tenure')
plt.xlabel('Tenure (months)')
plt.ylabel('Average Weekly Usage (hours)')
plt.show()


# Correlation
df[['avg_weekly_usage_hours', 'tenure_months']].corr()
```

## 4. Customer Retention Analysis

```python
# Churn Rate
churn_rate = (df['churn']==1).mean()*100
print(f"Churn Rate: {churn_rate:.2f}%")
```

```python
# Churn by Subscription plan
pd.crosstab(
    df['plan_type'],
    df['churn'].map({0:"Not Churned", 1:"Churned"}),
    normalize='index'
)*100
```

```python
# Churn and tenure
df.groupby('churn')['tenure_months'].agg(
    ['count', 'mean', 'median', 'max']
)
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    data=df,
    x='churn',
    y='tenure_months'
)
plt.title("Customer Tenure by Churn Status")
plt.xlabel('Churn')
plt.ylabel('Tenure (months)')
plt.show()
```

```python
# Usage vs churn
df.groupby('churn')['avg_weekly_usage_hours'].agg(
    ['count', 'min', 'mean', 'median', 'max']
)
```

```python
# Last login vs churn
df.groupby('churn')['last_login_days_ago'].agg(
    ['min', 'mean', 'median', 'max']
)
```

```python
# plot
plt.figure(figsize=(8,5))
sns.boxplot(
    data=df,
    x='churn',
    y='last_login_days_ago'
)
plt.title("Days Since Last Login by Churn Status")
plt.xlabel('Churn')
plt.ylabel('Days Since Last Login')

plt.show()
```

## 5. Revenue Analysis

```python
# Revenue by plan
display(Markdown("#### Revenue by Plan"))
display(df.groupby('plan_type')['monthly_fee'].agg(
    ['count', 'mean', 'median', 'sum']
))
```

```python
plt.figure(figsize=(5, 3))

revenue_by_plan = df.groupby('plan_type')['monthly_fee'].sum()

plt.pie(
    revenue_by_plan,
    labels=revenue_by_plan.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Monthly Revenue by Subscription Plan")
plt.show()
```

```python
# Revenue and Engagement
df.groupby('plan_type').agg({
    'monthly_fee': 'mean',
    'avg_weekly_usage_hours': 'mean'
})
```

## 6. Customer Support Analysis

```python
# Support Tickets vs Churn
df.groupby('churn')['support_tickets'].agg(
    ['count', 'mean', 'median']
)
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    data=df,
    x='churn',
    y='support_tickets'
)
plt.title("Support Tickets by Churn Status")
plt.xlabel('Churn')
plt.ylabel("Number of Support Tickets")
plt.show()
```

```python
# Support tickets by plan
display(Markdown("#### Support Ticket by Plan"))
display(df.groupby('plan_type')['support_tickets'].agg(
    ['count', 'mean', 'median']
))
```

```python
# Support Ticket vs engagement
display(Markdown("#### Support Ticket vs Engagement"))
display(df[['support_tickets','avg_weekly_usage_hours']].corr())
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df, x='support_tickets', y='avg_weekly_usage_hours', hue='churn'
)
plt.title('Support Tickets vs Product Usage')
plt.xlabel('Support Tickets')
plt.ylabel('Average Weekly Usage Hours')
plt.show()
```

## Data Visualization

```python
fig, ax = plt.subplots(2, 3, figsize=(18, 10))
# Chart 1 — Countplot
sns.countplot(
    data=df, x='churn', ax=ax[0, 0]
)
ax[0, 0].set_title('Customer Churn Distribution')
ax[0, 0].set_xlabel('Churn')
ax[0, 0].set_ylabel('Number of Customers')

# Chart 2 — Countplot
sns.countplot(
    data=df, x='plan_type', hue='churn', ax=ax[0, 1]
)
ax[0, 1].set_title('Churn by Subscription Plan')
ax[0, 1].set_xlabel('Subscription Plan')
ax[0, 1].set_ylabel('Number of Customers')

# Chart 3 — Boxplot
sns.boxplot(
    data=df, x='plan_type', y='avg_weekly_usage_hours', ax=ax[0, 2]
)
ax[0, 2].set_title('Weekly Usage by Subscription Plan')
ax[0, 2].set_xlabel('Subscription Plan')
ax[0, 2].set_ylabel('Average Weekly Usage Hours')

# Chart 4 — Scatterplot
sns.scatterplot(
    data=df, x='tenure_months', y='avg_weekly_usage_hours', hue='churn', ax=ax[1, 0]
)
ax[1, 0].set_title('Usage, Tenure and Churn')
ax[1, 0].set_xlabel('Tenure (months)')
ax[1, 0].set_ylabel('Average Weekly Usage Hours')

# Chart 5 — Barplot
sns.barplot(
    data=df, x='plan_type', y='monthly_fee', estimator='sum', ax=ax[1, 1]
)
ax[1, 1].set_title('Total Monthly Revenue by Plan')
ax[1, 1].set_xlabel('Subscription Plan')
ax[1, 1].set_ylabel('Monthly Revenue')

# Chart 6 — Boxplot
sns.boxplot(
    data=df, x='churn', y='last_login_days_ago',ax=ax[1, 2]
)
ax[1, 2].set_title('Customer Inactivity by Churn Status')
ax[1, 2].set_xlabel('Churn')
ax[1, 2].set_ylabel('Days Since Last Login')

plt.tight_layout()
plt.savefig('./7_Data_Visualization-1.png', bbox_inches='tight')
plt.show()
```

## Business Insights
- The overall churn rate is 57.32%, indicating that customer retention is an important business concern.
- Premium plan customers shows highest churn **(58.05%)** followed by basic plan customers **(57.85%)** and standard plan customers **(56%)**.Since the difference in only about 2 percentage points, churn appears to be relatively consistent across subscription plans.
- Customers who churned had an average weekly usage of 12.25 hours, compared with 13.75 hours among retained customers. This suggests that lower product engagement is associated with a higher likelihood of churn.
- Churned customers have a higher level of inactivity, with an average of 32.88 days and a median of 34 days since their last login, compared with an average of 26.15 days and a median of 22 days among retained customers. This suggests that prolonged inactivity may be an early warning sign of potential churn
- Premium customers contribute the largest share of total revenue at 54%, followed by Standard customers at 31% and Basic customers at 15%. Notably, Premium customers alone generate more revenue than the combined contribution of Standard and Basic customers (46%), highlighting the Premium segment's importance to overall revenue.
- Customers with higher numbers of support tickets show a higher/lower churn rate, suggesting that customer support interactions may be associated with retention outcomes.

## Business Recommendation
- **Recommendation 1**: Create an early churn-warning system <br>
  Monitor Customers who show:
  - increasing last_login_days_ago
  - low avg_weekly_usage_hours
  - multiple support tickets
  - payment failures
- **Recommendation 2**: Improve onboarding <br>
  Introduce a stronger onboarding program during the first few months, including tutorials, guided product setup, and proactive customer support.
- **Recommendation 3**: Target low-engagement customers.
